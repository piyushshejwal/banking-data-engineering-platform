```python
from sqlalchemy import text

for _, row in changed_df.iterrows():

    customer_id = int(row["customer_id"])
    new_status = row["customer_status_stage"]

    with engine.begin() as conn:

        # Get current customer record
        current = conn.execute(
            text("""
                SELECT
                    customer_number,
                    customer_name,
                    customer_type,
                    customer_status,
                    risk_category,
                    kyc_status,
                    branch_id,
                    city_id
                FROM dim_customer
                WHERE customer_id = :customer_id
                  AND is_current = TRUE
                LIMIT 1
            """),
            {"customer_id": customer_id}
        ).mappings().first()

        # If no current record exists, skip
        if current is None:
            print(f"Customer {customer_id}: no current record found")
            continue

        # Expire old version
        conn.execute(
            text("""
                UPDATE dim_customer
                SET effective_to = CURRENT_DATE,
                    is_current = FALSE
                WHERE customer_id = :customer_id
                  AND is_current = TRUE
            """),
            {"customer_id": customer_id}
        )

        # Insert new version
        conn.execute(
            text("""
                INSERT INTO dim_customer (
                    customer_id,
                    customer_number,
                    customer_name,
                    customer_type,
                    customer_status,
                    risk_category,
                    kyc_status,
                    branch_id,
                    city_id,
                    effective_from,
                    effective_to,
                    is_current
                )
                VALUES (
                    :customer_id,
                    :customer_number,
                    :customer_name,
                    :customer_type,
                    :customer_status,
                    :risk_category,
                    :kyc_status,
                    :branch_id,
                    :city_id,
                    CURRENT_DATE,
                    NULL,
                    TRUE
                )
            """),
            {
                "customer_id": customer_id,
                "customer_number": current["customer_number"],
                "customer_name": current["customer_name"],
                "customer_type": current["customer_type"],
                "customer_status": new_status,
                "risk_category": current["risk_category"],
                "kyc_status": current["kyc_status"],
                "branch_id": current["branch_id"],
                "city_id": current["city_id"]
            }
        )

        print(f"Customer {customer_id}: SCD2 updated")
```

```python
import pandas as pd

# 1. Read transactions from staging
transaction_df = pd.read_sql(
    """
    SELECT
        transaction_id,
        account_id,
        transaction_type_id,
        transaction_date,
        debit_amount,
        credit_amount,
        transaction_status
    FROM stg_transactions;
    """,
    engine_staging
)

# 2. Create transaction amount
transaction_df["amount"] = (
    transaction_df["debit_amount"].fillna(0)
    + transaction_df["credit_amount"].fillna(0)
)

# 3. Lookup customer_id from account_id
customer_account_lookup = pd.read_sql(
    """
    SELECT
        account_id,
        customer_id
    FROM stg_customer_accounts;
    """,
    engine_staging
)

transaction_df = transaction_df.merge(
    customer_account_lookup,
    on="account_id",
    how="left"
)

# 4. Lookup account_key
account_lookup = pd.read_sql(
    """
    SELECT
        account_id,
        account_key
    FROM dim_account
    WHERE is_current = TRUE;
    """,
    engine
)

transaction_df = transaction_df.merge(
    account_lookup,
    on="account_id",
    how="left"
)

# 5. Lookup customer_key
customer_lookup = pd.read_sql(
    """
    SELECT
        customer_id,
        customer_key
    FROM dim_customer
    WHERE is_current = TRUE;
    """,
    engine
)

transaction_df = transaction_df.merge(
    customer_lookup,
    on="customer_id",
    how="left"
)

# 6. Create date_key as YYYYMMDD
transaction_df["date_key"] = pd.to_datetime(
    transaction_df["transaction_date"]
).dt.strftime("%Y%m%d").astype(int)

# 7. Prepare final fact dataframe
fact_transaction_df = transaction_df[
    [
        "transaction_id",
        "customer_key",
        "account_key",
        "date_key",
        "transaction_date",
        "amount",
        "transaction_status"
    ]
].copy()

# transaction_type_key is left NULL because
# dim_transaction_type has not been loaded yet.
fact_transaction_df["transaction_type_key"] = None

# Put columns in warehouse order
fact_transaction_df = fact_transaction_df[
    [
        "transaction_id",
        "customer_key",
        "account_key",
        "transaction_type_key",
        "date_key",
        "transaction_date",
        "amount",
        "transaction_status"
    ]
]

# 8. Load into warehouse
fact_transaction_df.to_sql(
    "fact_transactions",
    con=engine,
    if_exists="append",
    index=False
)

print("fact_transactions loaded:", len(fact_transaction_df))
```

```python
import pandas as pd

# 1. Read payments from staging
payment_df = pd.read_sql(
    """
    SELECT
        payment_id,
        transaction_id,
        payment_date,
        amount,
        payment_status
    FROM stg_payments;
    """,
    engine_staging
)

# 2. LOOKUP: transaction_id -> transaction_key
transaction_lookup = pd.read_sql(
    """
    SELECT
        transaction_id,
        transaction_key
    FROM fact_transactions;
    """,
    engine
)

payment_df = payment_df.merge(
    transaction_lookup,
    on="transaction_id",
    how="left"
)

# 3. LOOKUP: payment_date -> date_key
payment_df["date_key"] = pd.to_datetime(
    payment_df["payment_date"]
).dt.strftime("%Y%m%d").astype(int)

# 4. payment_method_key unavailable for now
payment_df["payment_method_key"] = None

# 5. Prepare warehouse columns
fact_payment_df = payment_df[
    [
        "payment_id",
        "transaction_key",
        "payment_method_key",
        "date_key",
        "payment_date",
        "amount",
        "payment_status"
    ]
].copy()

# 6. Load
fact_payment_df.to_sql(
    "fact_payments",
    con=engine,
    if_exists="append",
    index=False
)

print("Payments loaded:", len(fact_payment_df))
print("Transaction lookup failures:",
      fact_payment_df["transaction_key"].isna().sum())
```
