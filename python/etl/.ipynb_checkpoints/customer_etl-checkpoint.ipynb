```python
import pandas as pd
import mysql.connector
```

```python
from config import MYSQL_CONFIG

config = MYSQL_CONFIG.copy()
config["database"] = "bank_staging"

conn = mysql.connector.connect(**config)

print("Connected successfully")
```

```python
query = """
SELECT *
FROM stg_customers;
"""

df = pd.read_sql(query, conn)

print(df.shape)
```
